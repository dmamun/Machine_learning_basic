```python
import numpy as np

# Student dataset
# Columns: Age, Math, English, Science

students = np.array([
    [20, 85, 78, 90],
    [21, 72, 88, 75],
    [19, 95, 91, 89],
    [22, 65, 70, 68],
    [20, 88, 84, 92]
])

print("Complete Dataset:")
print(students)


# -----------------------------------
# 1. Indexing: Specific element
# -----------------------------------

print("\n1. First student's Math mark:")
print(students[0, 1])


# -----------------------------------
# 2. Indexing: Specific row
# -----------------------------------

print("\n2. First student's complete data:")
print(students[0])


# -----------------------------------
# 3. Last row
# -----------------------------------

print("\n3. Last student's data:")
print(students[-1])


# -----------------------------------
# 4. First column (Age)
# -----------------------------------

print("\n4. All students' Age:")
print(students[:, 0])


# -----------------------------------
# 5. Second column (Math)
# -----------------------------------

print("\n5. All students' Math marks:")
print(students[:, 1])


# -----------------------------------
# 6. Slicing: First 3 students
# -----------------------------------

print("\n6. First 3 students:")
print(students[:3])


# -----------------------------------
# 7. Slicing: Math + English
# -----------------------------------

print("\n7. Math and English marks:")
print(students[:, 1:3])


# -----------------------------------
# 8. First 3 students + Math/English
# -----------------------------------

print("\n8. First 3 students' Math and English:")
print(students[:3, 1:3])

# 9. Students with Math > 80

print("\n9. Students with Math > 80:")
print(students[students[:, 1] > 80])

# 10. Students with Science > 85
print("\n10. Students with Science > 85:")
print(students[students[:, 3] > 85])
# 11. Students with Math > 80 AND English > 80
condition = (students[:, 1] > 80) & (students[:, 2] > 80)

print("\n11. Math > 80 AND English > 80:")
print(students[condition])
# 12. Reverse the rows
print("\n12. Dataset in reverse:")
print(students[::-1])
```

Numpy operation


```python
import numpy as np

# ==========================================
# 1. Create Arrays
# ==========================================

marks = np.array([80, 65, 90, 55, 75])

print("Marks:")
print(marks)


# ==========================================
# 2. Arithmetic Operations
# ==========================================

print("\n--- Arithmetic Operations ---")

print("Add 5:", marks + 5)
print("Subtract 5:", marks - 5)
print("Multiply by 2:", marks * 2)
print("Divide by 2:", marks / 2)
print("Square:", marks ** 2)


# ==========================================
# 3. Comparison Operations
# ==========================================

print("\n--- Comparison Operations ---")

print("Marks > 70:", marks > 70)
print("Marks < 70:", marks < 70)
print("Marks == 75:", marks == 75)


# ==========================================
# 4. Filtering
# ==========================================

print("\n--- Filtering ---")

passed_students = marks[marks >= 60]

print("Passed marks:", passed_students)

high_marks = marks[marks > 80]

print("Marks above 80:", high_marks)


# ==========================================
# 5. Aggregate Operations
# ==========================================

print("\n--- Statistics ---")

print("Total:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard Deviation:", np.std(marks))
print("Variance:", np.var(marks))


# ==========================================
# 6. Position of Min/Max
# ==========================================

print("\n--- Index Operations ---")

print("Index of minimum:", np.argmin(marks))
print("Index of maximum:", np.argmax(marks))


# ==========================================
# 7. Sorting
# ==========================================

print("\n--- Sorting ---")

print("Sorted:", np.sort(marks))

print("Sorting indices:", np.argsort(marks))


# ==========================================
# 8. 2D Array
# ==========================================

students = np.array([
    [80, 70, 90],
    [60, 75, 65],
    [90, 85, 95],
    [55, 60, 50]
])

print("\n--- 2D Array ---")

print(students)


# ==========================================
# 9. Axis Operations
# ==========================================

print("\n--- Axis Operations ---")

# Total marks of each student
print("Student-wise total:")
print(np.sum(students, axis=1))

# Average marks of each student
print("Student-wise average:")
print(np.mean(students, axis=1))

# Subject-wise total
print("Subject-wise total:")
print(np.sum(students, axis=0))

# Subject-wise average
print("Subject-wise average:")
print(np.mean(students, axis=0))


# ==========================================
# 10. Matrix Operations
# ==========================================

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("\n--- Matrix Operations ---")

# Element-wise multiplication
print("Element-wise multiplication:")
print(A * B)

# Matrix multiplication
print("Matrix multiplication:")
print(A @ B)

# Dot product
print("Dot product:")
print(np.dot(A, B))


# ==========================================
# 11. Mathematical Functions
# ==========================================

numbers = np.array([1, 4, 9, 16, 25])

print("\n--- Mathematical Functions ---")

print("Square root:", np.sqrt(numbers))
print("Square:", np.square(numbers))
print("Absolute:", np.abs(numbers))
print("Power:", np.power(numbers, 2))


# ==========================================
# 12. Concatenation
# ==========================================

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("\n--- Concatenation ---")

result = np.concatenate((a, b))

print(result)
```

```python
import numpy as np

# =========================================
# 1. Student Marks Dataset
# =========================================

marks = np.array([
    [80, 70, 90],
    [60, 75, 65],
    [90, 85, 95],
    [55, 60, 50]
])

print("Original Marks:")
print(marks)

print("Shape:", marks.shape)


# =========================================
# 2. Add 5 Marks to Everyone
# =========================================

new_marks = marks + 5

print("\nAfter Adding 5:")
print(new_marks)


# =========================================
# 3. Subject-wise Maximum
# =========================================

maximum = np.max(marks, axis=0)

print("\nSubject-wise Maximum:")
print(maximum)

print("Maximum Shape:", maximum.shape)


# =========================================
# 4. Normalize Marks
# =========================================

normalized = marks / maximum

print("\nNormalized Marks:")
print(normalized)


# =========================================
# 5. Subject-wise Mean
# =========================================

mean = np.mean(marks, axis=0)

print("\nSubject-wise Mean:")
print(mean)


# =========================================
# 6. Center the Data
# =========================================

centered = marks - mean

print("\nCentered Data:")
print(centered)


# =========================================
# 7. Different Weight for Each Subject
# =========================================

weights = np.array([0.30, 0.30, 0.40])

weighted_marks = marks * weights

print("\nWeighted Marks:")
print(weighted_marks)


# =========================================
# 8. Final Score
# =========================================

final_score = np.sum(weighted_marks, axis=1)

print("\nFinal Scores:")
print(final_score)
```

```python
import pandas as pd

data = {
    "Name": ["Rahim", "Karim", "Ayesha", "Sakib", "Nadia", "Hasan", "Tania", "Rafi"],
    "Age": [22, 25, 19, 30, 27, 21, 35, 24],
    "Height": [1.70, 1.75, 1.60, 1.80, 1.65, 1.72, 1.68, 1.78],
    "Weight": [65, 80, 55, 90, 60, 70, 85, 75],
    "Math": [85, 70, 90, 60, 88, 75, 65, 92],
    "English": [78, 65, 85, 55, 80, 70, 72, 88],
    "Science": [90, 72, 95, 58, 85, 80, 68, 94],
    "Study_Hours": [5, 3, 6, 2, 5, 4, 3, 6],
    "Attendance": [95, 70, 98, 60, 90, 85, 75, 97],
    "Monthly_Income": [30000, 45000, 25000, 60000, 40000, 28000, 55000, 50000],
    "Department": ["CSE", "EEE", "CSE", "BBA", "CSE", "EEE", "BBA", "CSE"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["BMI"] = df["Weight"] / (df["Height"] ** 2)
def bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"

df["BMI_Category"] = df["BMI"].apply(bmi_category)

print(df)
```

Merge and joint

inner merge and outer merge

```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Rahim", "Karim", "Ayesha", "Sakib"]
})

print(students)
```

```python
results = pd.DataFrame({
    "Student_ID": [103, 104, 105, 106],
    "Marks": [90, 65, 88, 75]
})

print(results)
```

```python
df_inner = pd.merge(
    students,
    results,
    on="Student_ID",
    how="inner"
)

print(df_inner)
```

```python
df_outer = pd.merge(
    students,
    results,
    on="Student_ID",
    how="outer"
)

print(df_outer)
```
